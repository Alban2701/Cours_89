# PEP 8

## Introduction

La **PEP 8** est une guide de style pour les codes pythons qui a pour objectif de standardiser l'écriture de ces derniers. Cela permet de rendre le code facilement lisible par tous les développeurs connaissant le langage. 

Par exemple regardez cette phrase : "bonjour , est ce Que vous pouvez m achETer ces ingrédient s il vous plaît:huile,cacao,  farine et oeufs".
La syntaxe est correcte, on peut lire la phrase et la comprendre, mais on est d'accord que ça fait mal aux yeux : 
- Il n'y a pas de majuscule en début de phrase ;
- La virgule doit être collé au mot précédent et avoir un espace avant le mot suivant ;
- Pas de trait d'union entre "est" et "ce" ;
- Il y a un mot commun qui porte une majuscule sur la première lettre ;
- Un verbe à l'infinitif qui a deux majuscule en plein milieu ;
- Pas d'apostrophe pour lier "me" et "acheter" ;
- Les deux-points (":") doivent être entourés d'espaces ;
- La phrase ne se termine pas par un point d'interrogation.

Donc en respectant la typographie et la grammaire on a : "Bonjour, est-ce que vous pouvez m'acheter ces ingrédients s'il vous plait : huile, cacao, farine et oeufs ?"

On va voir dans ce document les principales règles de la **PEP 8**. Il vous sera demandé de respecter ces règles dans tous les documents qu'il vous sera demandé de rendre.

Si vous utilisez des IDE (Environnement de Développement Intégré), certaines extensions vous permettent de formater votre code automatiquement afin de respecter les conventions. Ils sont pratique et vous êtes invités à les utiliser. CEPENDANT, il faut que vous gardiez ces règles en tête car vous serez évalués dessus.

Les exemples qui suivent ont majoritairement été extrait de la documentation officiel que vous pouvez trouver ici <https://peps.python.org/pep-0008/>.
J'ai essayé de traduire au mieux.

### Indentation et grandes lignes continues


#### Indentations et utilité

L'indentation a pour rôle d'indiquer une entrée dans un bloc de code (`for`, `while`, `if`, `match`, `with`), dans une définition de fonction, méthode ou classe, ou alors pour "casser" les grandes lignes afin que le code soit plus lisible.


Chaque niveau d'indentation doit être constitué de 4 espaces. (Pas de tabulation !)

```python
for i in range(5):
    print(i)
```

#### Exemples et comparaison de bonnes et mauvaises manières d'indenter

Indentation correcte :

```python

# On ajoute 4 espaces (un niveau d'indentation supérieur) pour distinguer les argument du reste du code.
def long_function_name(
        var_one, var_two, var_three,
        var_four):
    print(var_one)

# Aligné avec le premier caractere du premier argument.
foo = long_function_name(var_one, var_two,
                         var_three, var_four)

# Les indentation pendantes doivent ajouter un niveau d'indentation
foo = long_function_name(
    var_one, var_two,
    var_three, var_four)
```

Parmi les deux manières de déclarer `foo`, vous pouvez choisir celle que vous préférez.

Indentation incorrecte :
```python

# Les arguments sur la première ligne sont interdits lorsqu'on utilise l'alignement vertical par rapport à l'appel ou la déclaration.
foo = long_function_name(var_one, var_two,
    var_three, var_four)

# Les indentations liées au code de la fonction ne sont pas distinguables de la déclaration des arguments
def long_function_name(
    var_one, var_two, var_three,
    var_four):
    print(var_one)
```


#### Dans les blocs de codes

Lorsque la partie conditionnel d'une déclaration `if` ou d'un autre bloc est assez long pour nécessiter d'être écrit sur plusieurs lignes, il est judicieux d'ajouter deux niveaux d'indentations pour écrire la partie conditionnelle, ou bien d'utiliser un commentaire pour séparer la partie conditionnelle de la partie logique.

```python
# Pas d'indentation supplémentaire
if (this_is_one_thing and
    that_is_another_thing):
    do_something()  # Difficile de distinguer la partie logique de la partie conditionnelle

# Ajouter un commentaire permet de distinguer les deux parties
# dans les IDE ayant la coloration de la syntaxe
if (this_is_one_thing and
    that_is_another_thing):
    # Lorsque les deux sont vrais, on rentre dans la boucle
    do_something()

# Ajouter un niveau d'indentation supplémentaire dans la ligne continue conditionnelle.
if (this_is_one_thing
        and that_is_another_thing):
    do_something()
```

Lorqu'une ligne continue contient plusieurs opérateurs (`and`, `or`, `+`, `-`, ...) et que l'on casse la ligne, il vaut mieux mettre les opérateurs en début de ligne afin qu'ils se retrouvent sur la même colonne :

```python
# Incorrect:
# Les opérateurs se trouvent loin de leurs opérandes
income = (gross_wages +
          taxable_interest +
          (dividends - qualified_dividends) -
          ira_deduction -
          student_loan_interest)

# Correct:
# plus lisible, on voit mieux la corresponpdance opérateurs-opérandes
income = (gross_wages
          + taxable_interest
          + (dividends - qualified_dividends)
          - ira_deduction
          - student_loan_interest)
```

### Taille maximum d'une ligne

La pep 8 préconise une limite de toutes les lignes à 79 caractères.

Historiquement, c'est pour faire tenir la ligne sur un ancien moniteur, mais on peut se demander pourquoi cette règle est maintenue aujourd'hui.

Aujourd'hui, la PEP8 avance le fait que ça permette de garder le code bien lisible, et/ou de mettre deux fenêtres de codes côte à côte, ce qui est utile pour les review de code notamment.

De même, ça peut permettre aux personnes malvoyante d'agrandir la taille de la police sans avoir de saut de lignes non-controllé.

En effet, les "auto-wrapper" (outil qui permet de retourner à la ligne automatiquement, Alt+Z sur VSCode), ne prennent pas en compte les conventions et ne rendent pas le code lisible malgré que tout soit visible à l'horizontal.

En résumé, vous devez respecter dans votre code ce nombre de caractère maximal par ligne :
- 72 pour les docstring et commentaires
- 79 pour le reste.

### Lignes vides

En fonction de ce que vous écrivez, il faut rajouter des lignes vides dans votre code afin de l'aérer et de le rendre plus lisible.

Avant chaque définition de fonction ou de classes, vous devez ajouter deux lignes vides.

Avant chaque définition de méthodes dans une classe, vous devez laisser une ligne vide.

Dans l'écriture de code, vous pouvez rajouter des lignes vides entre différentes sections pour séparer les différentes parties logique.

### Imports de modules

#### Comment écrire des imports

Les imports doivent se faire de la sorte :

```python
import os
import sys
from subprocess import Popen, PIPE
```

Des imports de différents modules sur la même ligne est interdit !

```python
import sys, os
```

#### La position des imports

Les importations doivent toujours être en haut du fichier, juste après des commentaires ou docstring de modules et avant la déclaration de variables globales et constantes.

Il faut trier les importations selon cet ordre : 

- import des librairies standard
- import des librairies complémentaire, également appelées "third-party"
- import des librairies locales

Chaque groupe d'importation doit être séparé par des lignes vides (sauf si au tout début du fichier).

#### Imports absolus VS relatifs

Il est préconisé d'utiliser des imports absolus (qui viennent de la racine).

```python
import mypkg.sibling
from mypkg import sibling
from mypkg.sibling import example
```

Les imports relatifs ne sont autorisé que pour les librairies complexes (ce qui sera rarement le cas chez nous).

```py
from . import sibling
from .sibling import example
```

Les imports de librairies standard doivent toujours utiliser l'import absolue.

Les classes peuvent également être importée de cette manière :

```py
from myclass import MyClass
from foo.bar.yourclass import YourClass
```

Si il y a un conflit dans le namespace, on peut également faire comme ceci : 

```py
import myclass
import foo.bar.yourclass as anOtherClass

instance = myclass.MyClass()
```

#### Les imports sauvages

Les imports sauvages (`from <module> import *`) permettent d'importer tout élément se trouvant dans un module.
Ils sont interdits dans la plupart des cas car impossible de savoir pour un humain ou un bot de savoir ce qu'il y a dans le namespace.

Exemple de mauvais et bon imports:

```py
# Incorrect

from mypkg import *

a = alien_function()
```

```py
# Correct
import mypkg

a = mypkg.known_function()
```

### Les espaces

#### Éviter les espaces inutiles


##### Juste après un ouvrant et juste avant un fermant

```py
# Correct
spam(ham[1], {"eggs": 2})

# Incorrect
spam(ham[ 1 ], { "eggs": 2 } )
```

##### Entre une virgule finale et une fermeture de parenthese

```py
# Correct
foo = (0,)  # Ceci est un tuple singleton !

# Incorrect 
bar = (0, )
```

##### Juste avant une virgule, un point-virgule ou un deux-points

```py
# Correct
if x == 4: print(x, y); x, y = y, x

# Incorrect
if x == 4: print(x , y) ; x , y = y , x
```

##### Dans les slices (fragment de liste)

le deux-points doit être entouré du même nombre d'espace de chaque côté (zéro ou un)

```py
# Correct:
ham[1:9], ham[1:9:3], ham[:9:3], ham[1::3], ham[1:9:]
ham[lower:upper], ham[lower:upper:], ham[lower::step]
ham[lower+offset : upper+offset]
ham[: upper_fn(x) : step_fn(x)], ham[:: step_fn(x)]
ham[lower + offset : upper + offset]

# Incorrect:

ham[lower + offset:upper + offset]
ham[1: 9], ham[1 :9], ham[1:9 :3]
ham[lower : : step]
ham[ : upper]
```

##### Avant les appels et définitions de fonctions et les slices

```py
# Correct
spam(1)
dct["key"] = 1

# Incorrect
spam (1)
dct ["key"] = 1
```

#### Définition de fonctions

##### Annotations de fonctions

```py
# Correct
def munge(input: AnyStr): ...
def munge() -> PosInt: ...
```
Le deux-points se colle au paramètre et laisse un espace avant le type.
Le signe de retour `->` est entouré d'espace s'il est présent.
Le deux-points de déclaration de fonction se colle au dernier caractère non spatial et il y a un retour à la ligne.

```py
#Incorrect
def munge(input:AnyStr): ...
def munge()->PosInt: ...
```

##### Arguments nommés et valeur par défaut

Pas d'espace autour du symbole `=` lorqu'il indique un argument nommé ou une valeur par défaut pour une fonction qui n'est pas annotée.

```py
# Correct:
def complex(real, imag=0.0):
    return magic(r=real, i=imag)

# Incorrect:
def complex(real, imag = 0.0):
    return magic(r = real, i = imag)
```

##### Fonctions annotées et possédant des arguments nommés

En revanche, lorsqu'il y a des annotations de types et des arguments nommés, le symbole `=` est entouré d'espace

```py
# Correct:
def munge(sep: AnyStr = None): ...
def munge(input: AnyStr, sep: AnyStr = None, limit=1000): ...

# Incorrect
def munge(input: AnyStr=None): ...
def munge(input: AnyStr, sep:AnyStr, limit = 1000): ...
```

### Plusieurs expressions sur une ligne

C'est fortement déconseillé !

On préfère

```py
if foo == 'blah':
    do_blah_thing()
do_one()
do_two()
do_three()
```

plutôt que 

```py
if foo == 'blah': do_blah_thing()
do_one(); do_two(); do_three()
```

Donc il vous sera toujours demandé de faire une ligne par expression.

### Les virgules finales

Souvent optionnelles, elles sont surtout utilisées pour définir des tuples singleton (des tuples d'un seul élément).
Bien qu'ils peuvent être définis sans parenthèse, il faut les définir avec :

```py
# Correct
FILES = ("setup.cfg",)

# Incorrect
WRONG = "setup.cfg",
```

Pour définir des liste ou des séries d'arguments qui sont suscéptibles d'être agrandies, on peut également mettre des virgules finales, mais il faut faire un retour à la ligne avant de mettre le caractère fermant.

```py
# Correct:
FILES = [
    'setup.cfg',
    'tox.ini',
    ]
initialize(FILES,
           error=True,
           )

# Incorrect:
FILES = ['setup.cfg', 'tox.ini',]
initialize(FILES, error=True,)
```

### Commentaires

Il est tout d'abord préférable de ne pas mettre de commentaire plutôt que mettre des commentaires qui contre-disent le code.

Il est préconisé de faire des phrases complètes qui commencent par une majuscule (à moins que ce soit un indentifiant qui commence par une minuscule)

Dans un commentaire contenant plusieurs phrases, le point qui finit une phrase doit être suivi par un espace.

Il faut écrire les commentaire en anglais, sauf si vous êtes sûr à 120% qu'il ne sera lu par personne d'autre que quelqu'un qui parle la même langue que vous. (C'est pas moi qui le dit ! https://peps.python.org/pep-0008/#comments)

```py
# Voici une ligne de commentaire
a = 2

# Puis une autre```

#### Bloc de commentaires

Utilisé généralement pour décrire une partie ou tout le code qui le suit.

Il suit l'indentation du bloc de code dans lequel il se trouve.

Chaque ligne commence par un symbole `#` et un espace (sauf s'il y a du texte indenté). Les paragraphes sont séparés par une ligne de commentaire vide (qui commence quand même par `#`)

```python
# Ce code est vraiment compliqué
# Il fait des chose qu'aucun être humain est capable de comprendre
#
# Heureusement qu'il y a des blocs de code qui expliquent ce que ça fait

if a == 2:
    # Une condition !
    # Elle rend le code encore plus compliqué...
    #
    # On se fait tous remplacé par l'IA
```

##### Commentaire Inline

A utiliser avec parcimonie

Ils doivent être séparés de l'expression par au moins 2 espaces et commencent par un `#` et un espace.

Ils sont distrayant et ne doivent pas être utilisé pour décrire quelque chose d'évident.

Ne faites pas ça:
```py
update_screen()  # Mettre à jour l'écran
```

Ici, en revanche, c'est pratique:

```py
x += 5  # Cinq pixel de marge pour plus de visibilité. C'est mieux comme ça.
```

### String de documentation (Docstring)

Ils servent à documenter un module, une fonction, une methode. Ils doivent tous avoir une docstring, juste après leur définition.

Exemples :

```py

def foo_function():
    """Returns something incredible"""

    return "Something Incredible"

def another_function(nb: int | None = None) -> int | str:
    """This function does a weird thing.
    
    Parameters:
    - nb (int | None): a number the function can return (default None)

    Returns:
    - nb if given, "Hello World!" else.
    """

    if nb:
        return nb
    else:
        return "Hello World!"

```

On note que les caractères fermants la docstring sont sur leur propre ligne, sauf pour les docstring d'une seule ligne.

### Convention de nommage

#### Les différents styles de nommage

- variables: `snake_case`
- constants: `UPPERCASE`
- fonctions: `snake_case`
- classes: `PascalCase`

#### Compatibilité ASCII

Les identifiants ne doivent contenir que des caractères ASCII

#### Noms à éviter

Pour déclarer des variables à un seul caractère: ne jamais utiliser de caractères ambigües lors de déclaration de variable tel que 'l' (minuscule de L) qui peut être confondu avec '1', 'O' (la lettre o Majuscule) qui peut être confondue avec 0, et la lettre 'I' (i majuscule) qui peut parfois être confondue avec un 'L' minuscule.

#### Nom de classes

Le nom des classes doivent être en `PascalCase`

```py
class MyClass:
    pass
```

#### Nom de fonctions et de variables

Les fonctions et variables doivent être nommés en `snake_case` (chaque mot séparés par un underscore)

#### Arguments de méthodes et fonctions

Toujours utiliser `self` comme premier argument de méthode d'instance.

Si un nom de fonction déclenche un conflit avec un nom déjà pris, il se fait de rajouter un undescore après le nom plutôt que de le remplacer par une abréviation (`class_` est mieux que `clss`). Cependant, il est toujours préférable d'éviter ce genre de conflit.

#### Nom de méthodes et variables d'instances

Toujours utiliser la règle du `snake_case`.
Si la variable ou la méthode n'est pas publique, rajouter un underscore:

```py
class MyClass:
    """random class"""
    
    def __init__():
        var1 = 1
        _private_var = 2
```

#### Constantes

Les constantes sont écrites tout en majuscule avec des underscores séparant les mots: `TOTAL`, `MAX_OVERFLOW`.
