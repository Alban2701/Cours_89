# Boucles for / while et les types séquentiels

## Qu'est-ce qu'une boucle ?

### Définition

En programmation, une boucle est un bloc de code qui sera rejoué autant de fois qu'une condition est respectée.

Exemple dans la vie réelle :

- *"Tant qu'il n'est pas 17h30, je travaille"*
- *"Pour chaque logiciel sur mon ordinateur, je l'éteins"*
- *"Dix fois, je cours le tour d'un terrain"*

### Avantages

L'avantage d'une boucle en programmation, c'est qu'elle permet d'effectuer des actions répétées sans réécrire le code. On gagne donc du temps d'écriture et cela rend le code dynamique, qui s'adapte à la taille de données variables.

## Types de boucles

#### Boucle `while`

Le bloc de code situé dans une boucle `while` (qui se traduit par "tant que") est exécuté tant que l'expression après `while` est vérifiée.

C'est plus ou moins comme un bloc `if`, mais qui se réitère tant que la condition est vraie.

La syntaxe de base est la suivante : 

```py
while condition:
    # Code repete tant que la condition est True
```

1. Compteur simple

````python
max = 10
i = 0
while i <= 10:
    print(i)
    i += 1
````

2. Saisie utilisateur validée

````python
secret = "myUltim4tePassword!"
input_password = ""

while input_password != secret:
    input_password = input("Entrez le mot de passe")

print("Accès accordé")
````

3. Lecture de données jusqu'à une sentinelle:

````python
total = 0
nombre = -1
while nombre:  # équivaut à bool(0) => False
    nombre = int(input("Entrez un nombre (0 pour arrêter): "))
    print()
    total += nombre

print(f"Total: {total}")
````

La condition est évaluée avant chaque itération, et non pendant l'itération. Ainsi, si une valeur change impliquant que la condition ne soit plus vraie, le reste du bloc de code sera tout de même exécuté.

````python
i = 0
while i < 5:
    print(i)
    i += 1
````

````python
i = 0
while i < 5:
    i += 1
    print(i)
````

Il est impératif de déterminer une bonne porte de sortie pour une boucle while, sinon on risque la boucle infinie et votre ordinateur va vous détester poru ça.

````python
# /!\ ZONE DANGER

i = 0
while i <= 0:
    print(i)
    i -= 1
````

Python a quand même une protection et limite le nombre d'itérations (qui est tout de fois assez grand).

Si jamais ça vous arrive, votre seul ami sera le fameux `CTRL+C` ou la patience d'attendre que votre code s'interrompe.

#### Boucles `for`

La Boucles `for` permet d'itérer sur une séquence (liste, chaîne, range, tuple). Cela veut dire que pour chaque élément de cette séquence, le bloc de code sera exécuté.

Pour voir ce que sont les séquences : https://docs.python.org/fr/3.14/library/stdtypes.html#sequence-types-list-tuple-range

La syntaxe de base est la suivante

```python
for variable in sequence:
    # Code répété pour chaque élément dans le séquence
```

- `variable` prend successivement la chaque valeur de `sequence`
- `sequence` peut être une liste, une chaîne de caractère, ou d'autres iterables.

#### Exemples

##### Boucle for-range

`range` est une fonction Python  qui permet de créer une sequence à partir de 1, 2 ou 3 paramètres.
- `range(stop)`: 0 -> stop - 1
- `range(start, stop)` start -> stop - 1
- `range(start, stop, step)` start -> stop - 1 avec un pas de step

````python
for i in range(5):  # Un seul paramètre => fin de la range
    print(i)
````

````python
for i in range(2, 6):  # Deux paramètres => début et fin
    print(i)
````

````python
for i in range(0, 10, 2):  # Trois paramètres: début, fin, intervale
    print(i)
````

##### Boucle for sur liste

Les liste (`list`) sont un ensemble d'éléments entre crochets. Il peut y avoir un nombre infini d'objet de n'importe quel type.

````python
ue_countries = ["Allemagne", "Autriche", "Belgique", "Bulgarie", "Chypre", "Croatie", "Danemarque", "Espagne", "Estonie", "Finlande", "France", "Grèce", "Hongrie", "Irlande", "Italie", "Lettonie", "Lituanie", "Malte", "Pays-Bas", "Pologne", "Portugal", "Roumanie", "Slovaquie", "Slovénie", "Suède", "Tchéquie"]

for country in ue_countries:
    if country.lower().startswith(("a", "e", "i", "o", "u", "y")):
        print(country)
````

##### Boucle for range avec `len`

`len` est une fonction qui permet de récupérer la longuer d'une séquence.

````python
for pos in range(len(ue_countries[0:10])):
    print(f"{pos + 1}: {ue_countries[pos]}")
````

##### Boucle for sur `str`

````python
from string import punctuation, whitespace
vowels = []
consonants = []

message = "vers l'infini et au-dela"

for char in message:
    if char in "aeiouy":
        vowels.append(char)

    elif char not in punctuation + whitespace:
        consonants.append(char)        

print(f"Voyelles: {vowels}\nConsonnes: {consonants}")
````

##### Fonctions bonus

1. `enumerate` est une fonction qui prend en paramètre une sequence et optionnellement un paramètre `start` et permet de boucler sur cette séquence en récupérant l'id de la variable et la variable elle-même.

````python
for id, country in enumerate(ue_countries):
    if id % 3 == 0:
        print(f"{id}: {country}")
````

2. `zip` est une fonction qui itère sur pluseirus itérables en parallèle. Elle fait d'autres choses assez complèxes qu'on ne mentionnera pas ici.

````python
noms = ["Toto", "Tata", "Kylo", "Malo"]
ages = [16, 92, 35, 1]

for item in zip(noms, ages):
    nom = item[0]
    age = item[1]
    print(f"{item[0]} a {age} an{"s" if age > 1 else ""}")
````

## Les types séquentiels

Il existe 3 types séquentiels élémentaires: les listes (objet `list`), les n-uplets (objet `tuple`) et les intervalles (objet `range`). D'autres types plus spécifiques existent tels que les [séquences binaire](https://docs.python.org/fr/3.14/library/stdtypes.html#binaryseq) et les chaînes de caractères (`str`).

### Les listes

Une [liste](https://docs.python.org/fr/3.14/library/stdtypes.html#list) est un objet séquentiel mutable (c'est à dire qui peut être modifié) qui peut contenir un nombre indéfini d'objet de n'importe quel type.

Elle se présente sous la forme suivante :

````python
my_list = [1, "a"]
print(my_list)
````

Une liste se reconnaît grâce à ses crochets.
Il y a plusieurs manière de construire une liste :

- en utilisant une paire de crochets pour indiquer une liste vide : [] ;
- au moyen de crochets, en séparant les éléments par des virgules : [a], [a, b, c] ;
- en utilisant une liste en compréhension : [x for x in iterable] ;
- en utilisant le constructeur du type : list() ou list(iterable).

`my_list` est en réalité une référence à une liste stockée en mémoire.

````python
an_other_list = my_list  # an_other_list prend cette même référence
an_other_list.append("new item")
print(an_other_list)
print(my_list)
````

Puisque `my_list` et `an_other_list` pointe vers la même référence, les actions faites sur `my_list` se perçoivent aussi sur `an_other_list`.

C'est pourquoi, si on veut vraiment donner une liste identique à une autre variable et non la référence, il faut utiliser `list.copy()`

````python
an_other_list = my_list.copy()
an_other_list.append("final item")
print(an_other_list)
print(my_list)
````

````python
# On peut créer une liste à plusieurs dimensions

lists = [[] for i in range(3)]

for i, l1 in enumerate(lists):  # i correspond aux colonnes
    for j in range(3):  # j correspond aux lignes
        l1.append((i, j))

print(lists)  # On a un tableau en deux dimensions de 3 par 3
    
````

### Les N-uplets

Aussi appelés tuples, ce sont des objets séquentiels immuables. C'est-à-dire qu'ils ne peuvent pas être modifié. Ils sont utilsié générallement pour stocker des ensemble de données hétérogènes (comme des paires clé / valeurs)

Un tuple se présente sous la forme suivante :

````python
my_tuple = (1, "a")
print(my_tuple)
````

Les n-uplets peuvent être construits de différentes façons :

- en utilisant une paire de parenthèses pour désigner le n-uplet vide : () ;
- en utilisant une virgule, pour créer un n-uplet d'un élément : a, ou (a,) ;
- en séparant les éléments avec des virgules : a, b, c ou (a, b, c) ;
- en utilisant la fonction native tuple() : tuple() ou tuple(iterable).

Il faut noter que c'est la virgule qui fait le tuple, et non les parenthèses (sauf pour le tuple vide). Les parenthèse sont donc en théorie facultatives et permettent surtout d'éviter l'ambiguïté syntaxique. Par exemple f(a, b, c) =/= f((a, b, c)). A gauche c'est un appel de fonction avec 3 arguments, tandis qu'à droite c'est un appel de fonction avec 1 argument qui est un triplet.

### Les Ranges

Le type `range` représente une séquence immuable de nombre et est couramment utilisé pour, comme on l'a vu, itérer un certain nombre de fois dans les boucles `for`

Une range se présente sous la forme suivante:

````python
my_range = range(5)
print(my_range)
````

Une range peut être construite de plusieurs manières différentes:

- `range(stop)` -> tous les entiers de 0 à stop - 1
- `range(start, stop)` -> tous les entiers de start à stop - 1
- `range(start, stop, step)` -> tous les entiers de start à stop -1 avec un step

### Conversion de séquences

Il est possible de convertir les séquences entre elles sauf pour obtenir une range

````python
tuple_2 = tuple(my_list)
list_2 = list(my_tuple)
a = range(5)
list_3 = list(a)

print(tuple_2)
print(list_2)
print(list_3)
````

### Slices

Les slices (ou tranches) sont une manière de récupérer une partie d'une séquence. Il est possible de créer une slice avec la [syntaxe dédiée](https://docs.python.org/fr/3.14/reference/expressions.html#slicings).


````python
from string import ascii_lowercase as alpha

print(alpha)
print(alpha[::2])
print(alpha[:5])
print(alpha[5:])
print(alpha[5:15])
print(alpha[5:20:3])
````
