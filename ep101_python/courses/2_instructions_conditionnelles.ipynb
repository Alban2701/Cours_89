# Instructions conditionnelles

Lorsqu'on code un programme, on a souvent besoin d'exécuter du code différement selon des situations variables.

On veut peut-être tester si un nombre est positif et entier, si une chaîne de caractères est assez longue ou contient certains caractères (comme pour les mots de passe par exemple)... C'est la base de la programmation.

Exemple:

Lire le statut d'un fichier avant de fermer un programme.

-> L'utilisateur clique sur "Fermer le programme" -> Le fichier est-il sauvegardé ? Si oui, fermer le programme; si non, proposer de sauvegarder le fichier puis fermer le programme.

## Rappel sur les booléens

Pour comprendre comment fonctionnent les instructions conditionnelles, il faut d'abord se souvenir de ce qu'est un booléen.

Un booléen, ou `bool`, est une valeur binaire, c'est à dire qui n'a que deux possibilités de valeur. En l'occurence, une variable booléenne en python ne peut avoir que la valeur `True` ou la valeur `False`.

L'objectif d'un booléen est de dire si une affirmation est vraie ou fausse.

- "La moitié de 4 est 1." -> Faux !
- "4 est paire." -> Vrai !

En Python, on utilise les opérateurs de comparaisons pour obtenir des valeur booléennes:

```python
print(4 / 2 == 1)
print(4 % 2 == 0)
```

Rappels des principaux **opérateurs de comparaison** :

| opérateur | signification |
| -- | -- |
| `<` | est inférieur à |
| `<=` | est inférieur ou égal à |
| `==` | est strictement égal à |
| `>=` | est supérieur ou égal à |
| `>` | est supérieur à |
| `!=` | est différent de |

__/!\\__ Ne pas confondre l'opérateur de comparaison `==` et l'opérateur d'assignation `=`

On peut également appliquer les opérateurs logiques `not`, `and` et `or` :

```python
print(True and False)
print(True or False)
print(not True)
print(not False)
```

En prenant l'exemple sur la moitié de 4 :

```python
everthing_is_fine = 4 / 2 == 1 and 4 % 2 == 0
print(everthing_is_fine)
```

Pour plus de clarté, c'est bien de mettre parfois des parenthèses sur les différentes opérations.

Ça donnerait `everything_is_fine = (4 / 2 == 1) and (4 % 2 == 0)`

### Conversion en booléen

En Python, on peut tout convertir en booléen, ce qui est très pratique.

Admettons que l'on souhaite savoir si une liste est vide, si un utilisateur a rempli un champs, si une valeur est à 0...

```python
cart = ["eggs"]
wallet = 2.8
todo = "buy eggs"

print(bool(cart))
print(bool(wallet))
print(bool(todo))

# Simulons l'achat des oeufs...
cart.remove("eggs")
wallet -= 2.8
todo = ""

print(f"{cart}: {bool(cart)}")
print(f"{wallet}: {bool(wallet)}")
print(f"{todo}: {bool(todo)}")
```

## Premières instructions conditionnelles

Une instruction conditionnelle commence avec l'instruction `if` suivi de l'expression conditionnelle. On y met enfin le bloc de code à exécuter si la condition est respectée.

Ce bloc de code est **indenté**, cela permet à python de "comprendre" ce qui est dans le bloc conditionnel ou non.

```python
is_even = 4 % 2 == 0
if is_even:
    print("4 est pair")
    
if everthing_is_fine:
    print("Vous connaissez vos mathématiques !")
```

Analysons le code juste au dessus :
- Déclaration de la variable `is_even` dont on assigne le résultat de l'expression `4 % 2 == 0` qui renvoie `True`
- `if is_even` vérifie l'état de `is_even`. Ici, `is_even` vaut `True` donc on exécute la suite
- on met un petit message de confirmation
- on vérifie `everything_is_fine` dont on a vu plus haut qui valait `False`
- la condition n'est pas vérifiée et la suite n'est pas exécutée

## Et sinon ?

Si la condition dans un if n'est pas respectée, on peut entrer dans un autre bloc de code conditionnel `else`. `if`/`else` - `si`/`sinon`...

```python
if everthing_is_fine:
    print("Ne devrait pas être affiché")

else: 
    print("Il va falloir revoir vos bases !")
```

Le code qui se trouve dans le bloc `else` n'est joué que si le block if juste avant n'est pas joué. Donc peu importe le nombre de bloc `if` qui se trouve avant lui dans le code, un `else` n'est connecté qu'au `if` qui le précède

```python
if True:
    print(1)

if False:
    print(2)

else:
    print(3)
```

```python
if False:
    print(1)

if True:
    print(2)

else:
    print(3)
```

## Sinon Si ... ?

Pour permettre au `else` de regader plusieurs bloc `if`, on va utiliser `elif` (contraction pour else if). 

Un bloc `elif` ne s'active que si les `if` ou `elif` précédents n'ont pas été validés.

```python
note = 85
if note >= 90:
    print("Excellent")
elif note >= 70:
    print("Bien")
elif note >= 50:
    print("Passable")
else:
    print("À améliorer")
```

L'ordre des conditions est importante. Python évalue le code de haut en bas. Si la première condition est `True`, alors le reste ne sera pas évalué.

## Blocs conditionnels imbriqués

On peut imbriquer des blocs de code conditionnels pour architecturer son arbre.

```python
age = 25

if 0 < age < 18:
    print("Vous êtes mineur !")

    if age == 17:
        print("Bientôt la majorité !")

else:
    print("La majorité, c'est être libre !")
```

Il faut cependant préférer les blocs `elif` aux boucles `if` imbriqués pour avoir une meilleure lisibilité.

## Exemple de code conditionnel

Voici un code qui imite un questionnaire en fonction des réponses de l'utilisateur.

Vous pouvez lancer le code et répondre aux questions

```python
# Générer un questionnaire en fonction de la situation personnelle

age = input("Quel est votre âge ?")

if not age.isdecimal():
    print("Il faut entrer un âge valide !")

elif int(age) >= 18:
    driver_license = input("Avez-vous le permis ? [y/N]\n>>> ")  # \n affiche un retour à la ligne
    
    if driver_license.lower() == "y" or driver_license.lower() == "yes":
        has_car = input("Avez-vous une voiture ? [y/N]\n>>> ")
        
        if has_car.lower() == "y" or has_car.lower() == "yes":
            car_model = input("Quel est le modèle de votre voiture ?\n>>> ")
            if "peugeot" in car_model.lower():
                print("Vous aimez donc la marque lionne !")
            elif "renault" in car_model.lower():
                print("Avez vous essayé la Renault Zoé ?")
            elif "mercedes" in car_model.lower():
                print("La fiabilité allemande")
            else:
                print("Très bonne voiture !")
        
        else:
            project_buying_car = input("Projetez-vous d'acheter une voiture ? [y/N]\n>>> ")
            
            if project_buying_car.lower() == "y" or project_buying_car.lower() == "yes":
                budget = input("Quel est votre budget ?\n>>> ")
                
                if budget.isdecimal():
                    budget = int(budget)
                    if budget > 35000:
                        print("Sacré budget !")
                    
                    elif budget > 15000:
                        print("Vous prévoyez d'acheter une voiture neuve ?")
                        
                    elif budget > 8000:
                        print("Ce qu'il faut pour une bonne voiture d'occasion.")

                    else:
                        print("C'est un budget assez serré !")

                else:
                    print("Vous n'avez pas entré une valeur correcte pour le budget")
                    
            
    else:
        getting_license = input("Êtes-vous actuellement en train de passer le permis? [y/N]\n>>> ")
        if getting_license:
            print("Bon courage !")
        
        else:
            print("Vive les mobilités douces !")

else:
    print("Vous n'êtes pas concerné par le questionnaire")
        
```

### Explication du code

- Les `input` permettent au code de demander à l'utilisateur d'entrer une valeur
- On vérifie si la valeur donnée par l'utilisateur ne contient que des nombres avec `str.isdecimal()`
- On envoie des questions en fonction des réponses précédentes de l'utilisateur grâce aux `if` / `elif` / `else`

## Blocs conditionnels et Pep 8

| Règle | Numérotation |
| --- | --- |
| Un niveau d'indentation doit être composée de 4 espaces. Il est important de différencier espaces (ASCII 32) et tabulation (ASCII 09) | pep89.2.1 |
| Le deux points (`:`) indiquant le début d'un bloc indenté doit être collé au caractère non-spatial le précédant | pep89.2.2 |
| Si une expression conditionnelle est précédée par une instruction, il faut les espacer d'un saut de ligne pour plus de lisibilité | pep89.2.3 |
| Un bloc conditionnel *b* imbriqué dans un bloc conditionnel *a* doit toujours être justifié par du code qui nécessite *a* et non *b* | pep89.2.4 |

## Pour aller plus loin

Lorsque l'on veut tester une seule variable plusieurs fois, au lieu de faire des `elif` à tout va, on peut utiliser l'instruction `match`. 

```python
langue = "espagnol"

match langue:
    case "français":
        print("Bonjour, je m'appelle Toto")
    case "espagnol":
        print("Holà, me llamo Toto")
    case "anglais":
        print("Hello, my name is Toto")
    case _:
        print("unknown language -_-")
```
