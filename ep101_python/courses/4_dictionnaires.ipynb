# Dictionnaires

## Définition

Un dictionnaire (ou tableau de correspondance / mapping) est une strcture de données en Python qui fait correspondre des valeurs [hashable](https://docs.python.org/fr/3.14/glossary.html#term-hashable) (clé) à des objets arbitraires (valeurs). Dans un dictionnaire, les clé sont uniques.

Un objet hashable est un objet dont sa valeur ne change pas dans le temps, elle reste fixe. Ex:
- int
- str
- float
- tuple
...

Il faut voir le dictionnaire comme un annuaire téléphonique. La clé c'est le nom du contact et la valeur son numéro de téléphone.

On va souvent utiliser les dictionnaires pour stocker des données indexées par un identifiant, compter des occurences ou représenter des objets (un étudiant avec ses notes).

## Syntaxe et Création

### Syntaxe de base

```python
dico1 = {"first name": "Najib", "last name": "Al Awar"}
print(dico1)
```

### Méthode Constructeur

```python
dico2 = dict(first_name="Alban", last_name="Genta")
print(dico2)
```

### Liste de tuple

```python
dico3 = dict([("first name", "Joe"), ("last name", "Abi Raad")])
print(dico3)
```

### Dictionnaire vide

```python
dico4 = {}
print(dico4)
```

### Dictionnaire et la PEP8

### Accécer et modifier des données

- Pour obtenir une donnée venant d'un dictionnaire

```python
name = dico1.get("first name")  # Erreur si la clé n'existe pas
name = dico1["first name"]  # Même chose, syntaxe différente

# Une valeur par défaut peut être fournie pour éviter l'erreur
age = dico1.get("age", "unknown")

print(name)
print(age)
```

- Ajouter ou modifier une donnée

```python
dico1["age"] = 29
print(dico1)

dico1["age"] += 2
print(dico1)
```

- Suppression d'une données

```python
del dico1["age"]  # Supprime la paire. Erreur si clé absente
print(dico1)

# Supprimer et retourner la valeur. Renvoie la valeur par défaut si clé absente
value = dico1.pop("age", "unknown")
print(dico1, value, sep="\n")

dico1.clear()  # Vide totalement le dictionaire
print(dico1)


```

- Vérifier l'existence d'une clé

```python
if "last_name" in dico2:
    print("Here !")
```

## Itérer sur un dictionnaire

Il existe plusieurs manière d'itérer sur un dictionnaire en utilisant une boucle for

```python
for key in dico2:  # Iterer sur les clés
    print(key)

for key, value in dico2.items():  # Sur la paire (clé, valeur)
    print(f"my {key} is {value}")

for value in dico2.values():  # sur les valeurs
    print(value)
```
