## C'est quoi Python ?

Python est un langage de programmation (comme le C, le Rust, le Java...)
Plus précisément, python est un langage **interprété**. Pour vulgariser: il existe donc un interpréteur (qu'on a installé ensemble), qui va traduire le code écrit et exécuter les lignes du programme au fur et à mesure de la lecture.

Vous connaissez peut-être le HTML ou encore le Markdown... Ce ne sont pas des langages de programmation ! Ce sont des langage de balisage. Pour plus d'info : https://fr.wikipedia.org/wiki/Langage_de_balisage

### Règle de syntaxe

La manière d'écrire du python est conventionnée par la PEP 8.

Les PEPs (pour Python Enhancement Proposals) sont des documents qui fournissent des informations concernant Python à la communauté, ou décrivant de nouvelles fonctionnalités pour Python ou son environnement.

La PEP 8 est un document qui recense des règles d'écriture de code Python afin de le rendre lisible facilement par d'autres développeurs.

Bien que l'interpréteur peut comprendre le code même si la convention n'est pas suivie, il vaut toujours mieux la suivre.

C'est pourquoi, il y aura un chapitre du cours dédié à la PEP 8 et des rappels tout au long des autres chapitre. Il vous sera également demandé de suivre ces conventions car vous serez noté dessus.

En python on peut afficher du texte grâce à la fonction print: `print("Hello World!")`

### Déclarer une variable

En python, comme dans la majorité des langages de programmation, on peut déclarer des variables pour les réutiliser plus tard.

```python
mon_prenom = "toto"
print(mon_prenom)
```

`mon_prenom = "toto"` est ce qu'on appelle une affectation. On déclare une variable (ici `mon_prenom`), on rajoute le signe `=` pour spécifier que c'est une affectation, puis on ajoute la valeur souhaitée: `"toto"`. En Réalité, on peut mettre ce qu'on veut derrière le symbole `=` tant que c'est reconnu par python. On peut même stocker une fonction !

```python
afficher = print
afficher("toto")
```

#### Les types numériques

##### Int

Les **int** ou *integer* sont les types d'objets représentant des nombres entiers, qu'ils soient positif (42) ou négatifs (-6).
On peut effectuer toutes les opérations de bases sur ces nombres :

```python
print(2 + 3)  # Addition
print(-2 - 3)  # Soustraction
print(2 * -3)  # Multiplication (symbolisé par le symbole '*')
print(2 / 3)  # Division (symbolisé par le symbole '/')
```

Il existe également des **opérateurs arithmétiques**[^1] qui permettent de faire des calculs un peu plus poussés.

[^1]C'est comme ça que l'on appelle les fonctions de calcul symbolisées par les symboles +, -, *, /

```python
# Opérateur de puissance 
# (symbolisé par les symboles '**')
print(3 ** 2)  

# On peut aussi calculer la racine carré de cette façon
print(9 ** 0.5)  

# Opérateur de division euclidienne 
# (symbolisé par les symboles '//')
print(10 // 3 )

# Opérateur de reste de la division euclidienne
# également appelé 'modulo' (symbolisé par le symbole '%')
print(10 % 3)  
```

Notez la sortie des calculs `9 ** 0.5` et `2 / 3`. Elles sont légèrement différentes. On en parle bientôt !

Enfin, on peut aussi comparer la valeur de deux nombres entiers grâce aux **opérateurs de comparaison** suivants :

| opérateur | signification |
| -- | -- |
| `<` | est inférieur à |
| `<=` | est inférieur ou égal à |
| `==` | est strictement égal à |
| `>=` | est supérieur ou égal à |
| `>` | est supérieur à |
| `!=` | est différent de |

```python
print(1 > 2)
print(3 <= 1 + 2)
print(8 == 4)
print(8 >= 5)
print(5 > 5)
print(0 != 1)
```

NB:

Lorsque vous appliquez le symbole `-` pour un nombre négatif, il s'applique sur le résultat final de l'opération se trouvant à sa droite. De ce fait :

```python
print(-(3**2) == (-3) ** 2)  # -9 == 9
```

Dans la partie gauche de la comparaison, le `-` s'applique sur le résultat de l'opération `3 ** 2` tandis que dans la partie de droite, le `-` ne s'applique que sur le `3` car il est encapsulé dans des parenthèses.

##### Float



Les **float** sont des nombres décimaux, vulgairement appelés *nombres à virgules*. Ils ressemblent aux **int** mais sont caractérisé par la présence d'un point `.` séparant la partie entière de la partie décimale.

ATTENTION : C'est bien un **point** `.` qu'il faut utiliser et non une **virgule** `,`.

Là aussi, on peut effectuer toutes les opérations mentionnées plus tôt.

```python
print(2.6 + 3.1)  # Addition
print(1.7 - 3.3333)  # Soustraction
print(2.8 * 3)  # Multiplication
print(2.7 / 3)  # Division
print(102.25**2.89)  # Puissance
print(10 // 0.3)  # Division euclidienne
print(5.8 % 3)  # Modulo
print(3.0 > 2.9)  # Comparaison
```

Il y a également une écriture alternative des nombres flottants. En effet, lorsque la partie décimale est nulle, alors on peut écrire le nombre sans la partie décimale mais toujours avec un point.

```python
print(2.0 == 2.)  # Les deux valeurs sont identiques !
print(type(2.0))  # Et c'est bien un nombre flottant !
```

Bien que cela peut faire économiser un caractère de code. Pour une question de lisibilité, on va toujours préférer la notation complète `2.0`.

##### Opérations entre différents types numériques

On peut sans problème faire des opérations entre des **int** et des **float**.

Ainsi :

```python
print(2 + 2)  # Affiche un int
print(2 + 2.0)  # Affiche un float
print(2.0 + 2.0)  # Aussi un float !
print(2 == 2.0)  # Bah oui, ça vaut la même chose !
```

Dès lors qu'il y a un nombre flottant dans une opération arithmétique, la sortie sera également un nombre flottant. Mais il y a une méthode pour savoir si un flottant est un entier :

```python
a = 2.0
print(a.is_integer())
```

Python a cependant un léger défaut: il calcule mal les très petits nombres.

```python
print(10 / 3)
```

`10 / 3` devrait normalement renvoyer `3.333333...3` mais au lieu de ça, ça termine par un 5. C'est lié au fonctionnement de Python et il existe des librairies qui remédient à ça. Ce ne sera cependant que dans très rare cas que cela nous posera problème.

#### String (str)

Les **string** sont un type d'objet qui sont caractérisé par une chaîne de caractères. On les reconnaît car elles commencent et finissent par un double guillement `"` ou un simple guillemet `'`.

```python
print("Je suis une chaîne de caractères !")
```

Notamment utilisées pour manipuler ou afficher du texte, elles ont beaucoup de méthodes avec lesquelles on peut s'amuser et traiter ces données.
On peut également leur appliquer certains opérateurs comme pour les types numériques.

```python
print("toto" + "tata")
print("ti" * 3)
print("ti" == "ti")
print("abc" > "bcd")
print("an" in "Banane")
print(str(912))
# N'oubliez pas que vous pouvez faire vos propres tests dans les blocs de code !
```

Oh, un petit nouveau ? `in` est un opérateur d'appartenance et, dans le cas des **string**, permet de vérifier si une chaîne de caractère se trouve dans une autre chaîne.

On peut également récupérer une certaines partie de la **string** grâce aux crochets. On appelle ça des **slices**

ATTENTION: Le premier caractère de la chaîne est à l'indice **0**.

| a | b | c | d | e | f | g | h |
| - | - | - | - | - | - | - | - |
| 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |

```python
print("abcdefgh"[5])
print("abcdefgh"[2:6])
```

Les **string** ont également ce qu'on appelle des méthodes. Ce sont des fonctions qui s'appellent sur un objet. On l'a vu avec **float** pour vérifier s'il était un **integer** (`(0.2).is_integer`).

Par "ce sont des fonctions qui s'appellent ..." on entend qu'on peut faire un appel de fonction et/ou de méthode. `print` est la fonction et `print()` est son appel de fonction

Les méthodes des **string** sont particulièrement utiles. On les utilise beaucoup pour les classer, les modifier, reconnaître des patterns etc.

En voici les principales, on verra les exemples juste après :

| Méthode | Effet | Cas d'usage |
| --- | --- | --- |
| `.capitalize()` | Renvoie une copie de la chaîne avec son premier caractère en majuscule et le reste en minuscule | Utile pour formater une phrase ou un prénom. |
| `.count()` | Renvoie le nombre d'occurence d'une sous-chaîne de caractère dans la chaîne cible. | Vous avec compté le nombre de fois que vous dites "du coup" ? |
| `.endswith()` | Renvoie `True` si la chaîne se termine par la valeur mise en paramètre, sinon `False`. Il y a également la méthode `startswith()`| Utile notamment pour gérer les extensions de fichiers |
| `.find()` | Renvoie l'indice de la première position dans la chaîne où la valeur mise en paramètre est trouvée | Pratique si on cherche l'emplacement d'un mot en particulier, pour extraire la suite du texte par exemple. |
| `.format()` | Formate la chaîne avec les valeurs mises en paramètre. Elle en prend autant qu'on veut ! | Celui là, il sert dès qu'on veut mettre une variable à un endroit particulier dans une chaîne. Très utile ! |
| `.isdigit()` | Renvoie `True` si la chaîne ne contient que des chiffres | Pour extraire des nombres depuis des **string** sans lever d'erreur |
| `.lower()` | Renvoie une copie de la chaîne avec tous les caractères différentiables sur la casse convertis en minuscules. | Si on veut empêcher qu'on nom de fichier se retrouve avec des majuscules, par exemple |
| `.removesufix()` | Si la **string** se termine par la chaîne mise en paramètre et que celle-ci n'est pas vide, renvoie la chaîne sans ce suffixe | Idéal pour retirer les extensions de fichiers ou les unités de mesures |
| `.split()` | Renvoie une **liste** des mots de la chaîne en utilisant le séparateur mis en paramètre | Du parsing pur et simple |
| `.strip()` | Renvoie une copie de la **string** dont les caractères spatiaux initiaux et finaux sont supprimés. | Pour supprimer les espaces avant et après les caractères |


Vous pouvez trouver la liste complète avec les exemples ici <https://docs.python.org/fr/3.12/library/stdtypes.html#string-methods>

Maintenant, regardons ce que font ces méthodes

```python
print("hello world!".capitalize())
print("hello world!".count("l"))
print("my_new_song.txt".endswith("txt"))
print("banana split".find(" "))
print("banana split".split(" "))
print("Bonjour, j'ai {} ans".format(25))
print("458".isdigit())
print("Bonjour, j'ai 25 ans".isdigit())
print("Je Ne Me SeNs PaS tRèS bIeN".lower())
print("songs.txt".removesuffix(".txt"))
print("pomme, banane, haricots, lessive, pâtes, nocciolata".split(", "))
print("                 J'ai beaucoup d'espace !             ".strip())
```

Bon à savoir : Il y a une manière plus simple d'utiliser la méthode `.format()` grâce aux **f-string** ce qui rend le code plus léger et lisible :

```python
a = 4
b = 7
print(
    f"Un rectangle de {a}cm de longueur et {b}cm de largeur fait {a * b}cm^2"
)
```

Si vous voulez afficher des guillemets dans votre **string**, vous pouvez suivre les méthodes qui suivent

```python
print('Je commence ma "string" avec des singlequotes')
print('Ou alors j\'utilise des "backslash" (\\)')
```
