# Les Fonctions

## Définition

La fonction en python est l'une de vos plus grandes alliées.

fonction: "Suite d'instructions qui renvoie une valeur à son appelant. On peut lui passer des arguments qui pourront être utilisés dans le corps de la fonction." (<https://docs.python.org/fr/3.12/glossary.html#term-function>)

En gros, c'est du code que l'on enregistre et que l'on pourra appeler plus tard, autant de fois que l'on veut.

Elle permet notamment d'éviter de se répéter dans le code et de le découper en blocs logique.

## Définir une fonction

Une fonction simple se définie comme ce qui suit :

```python
def my_function():
    return "Hello World!"
```

Et voici une fonction avec des paramètres

```python
def mul(x, y):
    return x * y
```

On distingue plusieurs éléments:

| Élement | Utilité |
| --- | --- |
| `def` | Mot clé permettant à python de comprendre qu'on définit une nouvelle fonction |
| `my_function` | Le nom de la fonction. Doit être en snake_case et commencer par une lettre |
| les parenthèses `()` | On y met les paramètres et les arguments. Si on n'en a pas besoin, on laisse les parenthèses vides. |
| les deux points `:` | Signigie la fin de la signature de la fonction. On fait un retour à la ligne et une tabulation d'une taille de 4 espaces |
| le code de la fonction | On peut y mettre toutes les opérations qui sont nécessaire au bon fonctionnement de notre programme |
| `return x` | (Optionnel) Mot clé pour renvoyer une valeur (ici x). Interrompt le cours de la fonction et le code situé après n'est jamais exécuté |

### Définition de fonction et Pep 8

Pour définir une fonction, il faut respecter ces règles :
- le nom de la fonction doit être en `snake_case`
- pas d'espace après la parenthèse ouvrante, ni avant la la parenthèse fermante
- les paramètres sont séparés par une virgule. Elle est adjacente au paramètre qu'elle suit et précède un espace
- les deux-points sont juste après la parenthèse fermante
- l'indentation est de 4 espaces

## Appeler une fonction

Une fonction n'est pas exécutée tant qu'elle n'est pas appelée.

Pour appeler une fonction, il faut mettre son nom, et y ajouter des parenthèses.

```python
var = my_function()
print(var)
```

La valeur de retour de `my_function` est stockée dans `var`, puis on affiche le résultat.

```python
result = mul(3, 8)
print(result)
```

Que se passe-t-il si on ne mets pas d'arguments à l'appel de la fonction `mul` ?

```python
result = mul()
print(result)
```

Python nous indique une error: il manque les arguements x et y !

### Différence entre paramètre et arguments

Les paramètres sont les entités nommées dans la définition d'une fonction qui décrivent un ou des arguments que la fonction accepte.

Voir <https://docs.python.org/fr/3.14/glossary.html#term-parameter>

## La portée d'une variable

Les variables ont ce qu'on appelle une portée. Une variable définie dans une fonction n'en sort jamais par défaut. Il faut la retourner ou la définir de manière global.

```python
def set_my_name():
    my_name = "tata"  # Défini dans le namespace de la fonction
    print(my_name)

set_my_name()
print(my_name)  # Erreur : 'my_name' n'est pas défini
```

On peut toujours récupérer une variable déclarée au niveau global pour l'afficher bien que ce ne soit pas une bonne pratique.

En revanche, on ne peut pas réassigner cette valeur.


```python
x = 2

def print_x():
    print(x)

def add_x():
    x = x + 2
    print(x)
    

print_x()
print(add_x())
```

```python
x = "global"

def local_function():
    x = "local"
    print(x)

local_function()
print(x)
```

Bien que la première définition de x est au niveau global, python va toujours préféré utiliser la variable au niveau local.

### Mot-clé `global`

```python
count = 0

def increment():
    global count
    count += 1

increment()
increment()
print(count)
```

Le mot clé global permet à Python de comprendre que count est une global variable et qu'à chaque fois qu'elle est utilisée dans la fonction, il doit aller récupérer sa valeur dans le namespace global.
Utiliser cette méthode n'est pas une bonne pratique car, bien qu'ici ce soit facile à suivre, on peut vite s'y perdre dans des codes plus long et complexe. On va donc préférer passer count en paramètre de la fonction.

```python
count = 0

def increment(x):
    x += 1
    return x

count = increment(count)
print(count)
```

## Documenter sa fonction

Au fur et à mesure que vous écrirez du code, vous oublierez peut-être l'utilité d'une fonction que vous avez écrit il y a un bout de temps. Ou encore, vous travaillez avec quelqu'un d'autre sur le projet et il a besoin de comprendre ce que fait votre code. 

Il y a donc deux manières complémentaires de documenter une fonction : le type hinting et les docstring.

### Le type hinting

Le type hinting consiste à indiquer les types des arguments attendus dans la fonction.

```python
def add_three(x: int):
    return x + 3

x: int = 3  # On peut aussi typer des variables lors de leur assignation
```

`add_three` attend en paramètre une valeur `x` qui est un int. Ici, c'est surtout utile lorsqu'on lit directement la définition de la fonction. Mais dans les éditeurs de code tels que VSCode, lorsqu'on fait appel à une fonction, l'éditeur nous montre sa définition et le type des paramètres attendus. Certaines extensions (ex. Pylance) se servent aussi de ce type hinting pour nous avertir (par exemple en soulignant l'erreur) si on utilise un variable qui n'est pas du bon type.

Pour plus de libertés et de possibilités dans le typage de votre code, vous pouvez utiliser la lirairie `typing`. Plus d’info: <https://docs.python.org/3/library/typing.html#module-typing>

```python
from typing import Optional, List, Dict

a_list: List[str] = ["can", "contain", "only", "str", "object"]
a_dict: Dict[str, Optional[int | str]] = {
    "France": 66_751_559, 
    "Luxembourg": 687_819,
    "Mars": None
    }
```

### Les Docstrings

Les doctrings sont l'équivalent des commentaires qui peuvent être écrites sur plusieurs lignes. Elles sont là pour expliquer ce que fait une fonction.

```python
def add_three(x: int):
    """Returns x + 3"""
    return x + 3
```

Cette docstring est assez simple, mais on peut la rendre plus complète et explicite si la fonction a besoin d'être plus détaillée.

```python
def euclidian_division(x: int, y: str) -> tuple:
    """Divide de the dividend by the divisor.
    
    Params :
      - x (int): the divident
      - y (int): the divisor

    Returns (quotient: int, remainder: int): the quotient and the remainder
    """
    quotient = x // y
    remainder = x - quotient
    return quotient, remainder
```

Avec cette docstring, la personne qui lit le code a toute les informations dont il a besoin. Il sait quels sont les arguments attendus lors de l'appel de fonction, ce qui se passe dans le code et ce qu'il va recevoir en retour.

## La récursivité

La récursivité est un principe en programmation qui implique d'appeler une fonction dans sa propre définition. Ce type de fonction est notamment utilisé pour des fonctions mathématiques. De la même manière que les boucles `while`, il faut toujours fare attention à avoit une porte de sortie

```python
def facto(n: int) -> int:
    """
    Returns n!
    """
    if n <= 1:
        return 1
    else:
        return n * facto(n - 1)

print(facto(6))
```

## Les fonctions et la pep8

Voici les règles à respecter dans le cadre d’écriture des fonctions.

| Règles | Numérotation |
| --- | --- |
| Il ne doit pas y avoir d'espace juste après un caractère ouvrant (`(`, `{`, `[`) | pep89.1.5 |
| Il ne doit pas y avoir d'espace juste avant un caractère fermant (`)`, `}`, `]`) | pep89.1.6 |
| Les déclarations, définitions de fonctions et identifiants, ainsi que les commentaires et docstring doivent être en anglais | pep89.1.11 |
| Les variables et noms de fonctions doivent être en **snake_case** | pep89.1.12 |
| Niveau d'indentation marqué de 4 espaces | pep89.2.1 |
| Le deux points indiquant le début d'un bloc indenté doit être collé au caractère non-spatial le précédent | pep89.2.2 |
| Toutes fonctions doivent avoir une docstring (simple ou complète en fonction de la complexité de la fonction). | pep89.5.1 |
| Toutes définitions doit être typées | pep89.5.2 |
